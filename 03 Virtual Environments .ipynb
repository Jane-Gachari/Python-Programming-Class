**PYTHON VIRTUAL ENVIRONMENTS**

A virtual environment is an isolated Python environment that allows you to install and manage packages for a specific project.

Each project can have its own:Python packages; package versions; dependencies.

This means that installing a package in one project does not automatically affect another project's environment.

Package - A reusable code that provides functionality you can use in your programs.

Dependencies - Are external tools, packages, or pre-written code libraries that your Python project relies on ("depends on") in order to work properly.

Instead of writing complex code from scratch for every task, you install and import code written by other developers. Because your code cannot run without those extra pieces, your project depends on them.

**Note:** Python  comes with built-in tools like ***math*** or ***random*** that are already included when you install Python—usually, when we say "dependencies," we mean extra packages that have to be installed separately using a command like ***pip install pandas***.

**Key rationale behind virtual environments:** One project, one isolated (customized) environment.

**Why are Python virtual environments so important?**

- Avoid dependency conflicts - when working on more than one project 
- Different projects can use different package versions seamlessly
- Keep projects remain organized - even when you're working on several  concurrently
- Each project maintains its own dependencies
- Improve reproducibility ; other developers can install the same dependencies
- Simplifies collaboration ; team members can recreate your project environment.
- Protect your global Python installation ; you avoid installing every package globally.

Examples of packages: numpy,Pandas,Matplotlib,Seaborn,Scikit-learn.

To install packages, we can use ***pip***, Python's package installer.

PS, the specific packages (or dependencies) are installed inside your project's virtual environment.
