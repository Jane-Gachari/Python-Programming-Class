**VERSION CONTROL & COLLABORATION ENVIRONMENT**

-A version control is a system or tool that helps to track changes I or others make on files over time. 

**Git & GitHub**
-Git is a version control system (tool) that allows me to track changes (or rather reconstruct events) in my file.

With Git, the following is possible:

- Track changes to a line of code
- Save important versions of a project
- Compare different versions of a project
- Create separate branches to experiment safely
- Combine changes to a file from different contributors

Note: Git works locally on the computer, even without an internet connection.

**GitHub** is an online platform which hosts your Git repositories.Using GitHub, the following is possible:

- Store projects online
- Share your work/projects portfolio with others (for instance tutors)
- Collaborate with other developers on an active project
- Review and discuss code with other developers (for instance, to identify a bug/error)
- Showcase your projects in a portfolio (for instance to potential employers, collaborators)

**(Key differences):** Git is the tool that tracks changes, while GitHub is the platform where you can host, share and collaborate on those projects.

**Git Terminology**

The following are concepts that help to logically understand how Git works:

Repository: A project folder whose history is tracked by Git.
- Working directory: The files you are currently working on.
- Staging area: An area where you select changes to include in your next commit.
- Commit: A snapshot of selected changes.
- Branch: A separate line of development.
- Remote repository: A repository hosted somewhere other than your local computer, such as GitHub.
- Clone: Creating a local copy of an existing remote repository.
- Push: Uploading local commits to a remote repository.
- Pull: Fetching and integrating changes from a remote repository.

**Git Workflow**

In practice, Git follows a specific process flow when you save changes;

- Working directory: Make changes to your files.
- Staging area: Select the changes you want to save.
- Local repository: Commit the selected changes.
- Remote repository: Push your commits to GitHub.

The workflow: *Working Directory → Staging Area → Local Repository → GitHub*

**Note:** Saving a file is not the same as committing it, and committing a file is not the same as pushing it to GitHub.

A GitHub repository is characterised by the following key elements:
- Repository name
- Repository description
- README.md file
- Branch information
- Commit history
- Code button

Note: A README introduces your project and explains what someone will find inside it.

GitHub allows you to create repositories with different visibility settings.

Public Repository: anyone can view it;useful for showcasing portfolios; can be shared via a link.
Private Repository: only authorized users can access it;useful for restricted projects; access must be granted.

**Note:**  Never upload passwords, API keys, confidential information or private datasets to a public repository.
