INTRODUCTION TO TERMINAL COMMANDS

A terminal is like a window frame which allows us to navigate and communicate with a computer (by typing in commands) when writing a program or when using a version control (for example Git) to work on a project.

A terminal allows you to execute commands to perform the following tasks:

- Navigate between folders
- Create files and directories
- View existing files
- Run Python programs
- Install packages
- Manage projects

Instead of clicking through different folders, menus etc, you can issue an instruction (command) for the computer to execute a task.

*For example: *

Instead of manually creating a folder, you can type:

*mkdir my-project*

Your computer interprets the command and creates the folder.

**Important terminology**

- Terminal: The application where you enter commands
- Command: An instruction given to the computer
- Directory: Another name for a folder
- Path: The location of a file or directory
